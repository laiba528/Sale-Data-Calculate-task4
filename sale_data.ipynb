```python
# manually sale data create
sale_data=[2300,1200,1500,1800,2000,2800,3000,3200,3467,2512,3478,9900,6593,9391,10000,9999]
print("sale_data",sale_data)
```

```python
# calculate number for max,min,sum
total_sale=sum(sale_data)    # calculute total sale
print("total_sale",total_sale)
high_sale=max(sale_data)       # calculute max sale
print("high_sale",high_sale)
lower_sale=min(sale_data)      # calculute min sale
print("lower_sale",lower_sale)
number_of_sale=len(sale_data)   # calculute number of sale
print("number_of_sale",number_of_sale)
avg_sale=total_sale/number_of_sale   # calculute avg of avg
print("avg_sale",avg_sale)
```

```python
if avg_sale > 4000 :
 print("sale perform is good")
else:
  print("sale perform is not good")
```
